```python
import numpy as np
import pandas as pd
```

**Overview of Dataset**

```python
o=pd.read_csv('WA_Fn-UseC_-Telco-Customer-Churn.csv')
o.head()
```

```python
o.info()
```

**EMPTY VALUES IN TOTALCHARGES COLUMN**

```python
j=o['TotalCharges']==" "
o[j]
```

**REMOVAL OF NON EMPTY COLUMNS**

```python
k=o['TotalCharges']!=" "
o=o[k]
```

**COVERSION OF DATA TYPE**

```python
o['TotalCharges']=o['TotalCharges'].astype(float)
o.info()
```

```python
o=o.set_index('customerID')
o
```

```python
o['Churn'].value_counts()
```

**CONVERSION OF BINARY (YES AND NO) TO BINARY(0 AND 1)**

```python
o['Churn']=o['Churn'].map({'Yes':1,'No':0})
o.head()
```

```python
bc=['Partner','Dependents','PaperlessBilling']
for e in bc:
    o[e]=o[e].map({'Yes':1,'No':0})
o.head()
```

```python
o['PhoneService']=o['PhoneService'].map({'Yes':1,'No':0})
o.head()
```

```python
o['gender']=o['gender'].map({'Male':1,'Female':0})
o.head()
```

**CONVERSION OF STRING DATA TO NUMERIC DATA(ONE HOT ENCODING)**

```python
jh=['MultipleLines','InternetService','OnlineSecurity','OnlineBackup','DeviceProtection','TechSupport','StreamingTV','StreamingMovies','Contract','PaymentMethod']
o=pd.get_dummies(o,columns=jh,drop_first=True)
```

```python
o.head()
```

**CONVERSION OF BOOLEAN DATA TO NUMERIC**

```python
o=o.astype({c:int for c in o.select_dtypes('bool').columns})
o.head()
```

```python
x=o.drop('Churn',axis=1)
y=o['Churn']
```

**SPLITTING DATA FOR TRAINING AND TESTING**

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
y_test.shape
```

**SCALING THE DATA**

```python
from sklearn.preprocessing import StandardScaler
sc=StandardScaler()
x_train=sc.fit_transform(x_train)
x_test=sc.transform(x_test)
```

**FITTING THE MODEL**

```python
from sklearn.linear_model import LogisticRegression
mo=LogisticRegression()
mo.fit(x_train,y_train)
```

**PREDICTING THE VALUES**

```python
y_pred=mo.predict(x_test)
```

```python
y_test
```

**METRICS OF THE MODEL**

```python
from sklearn.metrics import accuracy_score,precision_score,recall_score,confusion_matrix,classification_report
"accuracy score: ",accuracy_score(y_pred,y_test),'\n','Precision score: ',precision_score(y_pred,y_test),'\n','recall_score: ',recall_score(y_pred,y_test)
```

**CONFUSION MATRIX**

```python
import seaborn as sn
import matplotlib.pyplot as plt
cm=confusion_matrix(y_pred,y_test)
sn.heatmap(cm,annot=True,fmt='d',cmap='Blues',xticklabels=['No Churn','Churn'],yticklabels=['No churn','Churn'])
plt.xlabel('Predicted')
plt.ylabel('Actual')
plt.show()
```

**ANOTHER MODEL**

```python
from sklearn.ensemble import RandomForestClassifier
rc=RandomForestClassifier()
rc.fit(x_train,y_train)
y_pred2=rc.predict(x_test)
"accuracy score: ",accuracy_score(y_pred2,y_test),'\n' ,"precision score: ",precision_score(y_pred2,y_test),'\n','recall_score: ',recall_score(y_pred2,y_test)
```

```python
imp=pd.Series(rc.feature_importances_,index=x.columns).sort_values(ascending=False)
imp.head(10).plot(kind='barh')
plt.title("Top 10 Features")
plt.gca().invert_yaxis()
plt.show()
```
