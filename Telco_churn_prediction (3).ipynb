**TELECOMMUNICATION CHURN PREDICTION**<br>
Based on the dataset this model predicts whether the customer would churn or not

```python
import numpy as np
import pandas as pd
```

**Overview of Dataset**

```python
o=pd.read_csv('WA_Fn-UseC_-Telco-Customer-Churn.csv')
o.head()
```

```python
o.info()
```

**EMPTY VALUES IN TOTALCHARGES COLUMN**<br>
There were 11 rows in which Total Charges value was empty. Hence they were removed

```python
j=o['TotalCharges']==" "
o[j]
```

**REMOVAL OF ROWS WITH EMPTY VALUES**

```python
k=o['TotalCharges']!=" "
o=o[k]
```

**CONVERSION OF DATA TYPE**
In the dataset the Total Charges data type was string initially.It is changed to float

```python
o['TotalCharges']=o['TotalCharges'].astype(float)
o.info()
```

```python
o=o.set_index('customerID')
o
```

```python
o['Churn'].value_counts()
```

**CONVERSION OF BINARY (YES AND NO) TO BINARY(0 AND 1)**<br>
Few columns in this dataset have binary yes and no values but they were in string format.Hence they need to be converted to numeric form

```python
o['Churn']=o['Churn'].map({'Yes':1,'No':0})
o.head()
```

```python
bc=['Partner','Dependents','PaperlessBilling']
for e in bc:
    o[e]=o[e].map({'Yes':1,'No':0})
o.head()
```

```python
o['PhoneService']=o['PhoneService'].map({'Yes':1,'No':0})
o.head()
```

```python
o['gender']=o['gender'].map({'Male':1,'Female':0})
o.head()
```

**CONVERSION OF STRING DATA TO NUMERIC DATA(ONE HOT ENCODING)**<br>
In these particular columns the data in each column can be classified into more than 2 categories hence one hot encoding is used to convert them to numeric form

```python
jh=['MultipleLines','InternetService','OnlineSecurity','OnlineBackup','DeviceProtection','TechSupport','StreamingTV','StreamingMovies','Contract','PaymentMethod']
o=pd.get_dummies(o,columns=jh,drop_first=True)
```

```python
o.head()
```

**CONVERSION OF BOOLEAN DATA TO NUMERIC**<br>
After one hot encoding the data of those columns are in boolean form so we need to convert it to numeric form

```python
o=o.astype({c:int for c in o.select_dtypes('bool').columns})
o.head()
```

```python
x=o.drop('Churn',axis=1)
y=o['Churn']
```

**SPLITTING DATA FOR TRAINING AND TESTING**<br>
Now after Data preprocessing the data has been split for training and testing

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
y_test.shape
```

**SCALING THE DATA**<br>
Data has been scaled so that regression can happen easily

```python
from sklearn.preprocessing import StandardScaler
sc=StandardScaler()
x_train=sc.fit_transform(x_train)
x_test=sc.transform(x_test)
```

**FITTING THE MODEL**<br>
The Model used for prediction has been deployed here

```python
from sklearn.linear_model import LogisticRegression
mo=LogisticRegression()
mo.fit(x_train,y_train)
```

**PREDICTING THE VALUES**<br>
Here it predicts the values for the test data

```python
y_pred=mo.predict(x_test)
y_pred
```

```python
y_test
```

**METRICS OF THE MODEL**<br>
The essential metrics of the data are measured here (accuracy,precision,recall score)

```python
from sklearn.metrics import accuracy_score,precision_score,recall_score,confusion_matrix
"accuracy score: ",accuracy_score(y_test,y_pred),'\n','Precision score: ',precision_score(y_test,y_pred),'\n','recall_score: ',recall_score(y_test,y_pred)
```

**CONFUSION MATRIX**<br>
The confusion matrix have been plotted

```python
import seaborn as sn
import matplotlib.pyplot as plt
cm=confusion_matrix(y_test,y_pred)
sn.heatmap(cm,annot=True,fmt='d',cmap='Blues',xticklabels=['No Churn','Churn'],yticklabels=['No churn','Churn'])
plt.xlabel('Predicted')
plt.ylabel('Actual')
plt.show()
```

**ANOTHER MODEL**<br>
Another model randomforest classifier is used to compare the previous model 

```python
from sklearn.ensemble import RandomForestClassifier
rc=RandomForestClassifier()
rc.fit(x_train,y_train)
y_pred2=rc.predict(x_test)
"accuracy score: ",accuracy_score(y_test,y_pred2),'\n' ,"precision score: ",precision_score(y_test,y_pred2),'\n','recall_score: ',recall_score(y_test,y_pred2)
```

**Important features**<br>
Features on which churn depends the most

```python
imp=pd.Series(rc.feature_importances_,index=x.columns).sort_values(ascending=False)
imp.head(10).plot(kind='barh')
plt.title("Top 10 Features")
plt.gca().invert_yaxis()
plt.show()
```
